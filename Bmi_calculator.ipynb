```python
Name = input("Enter your Name : ")
Weight = int (input("Enter your current weight in pounds : "))
Height = int (input ("Enter your height in inches : "))

BMI=(Weight * 703)/(Height  * Height)
print(BMI)
```

```python
if BMI < 18.5:
  print(Name +" " +  "You are underweight ")
  print("The Health Risk is Minimal")
elif BMI >= 18.5 and BMI<=24.9:
  print(Name +" " + "Normal Weight")
  print("The Health Risk is Minimal")
elif BMI >=25 and BMI<=29.9:
  print(Name +" " +  "OverWeight")
  print("The Health Risk is Increased")
elif BMI >=30 and BMI<= 34.9:
  print( Name +" " +  "Obesity")
  print("The Health Risk is High")
elif BMI >= 35 and BMI<= 39.9:
  print(Name +" " +  "Severly Obese")
  print("The Health Risk is Very High")
elif BMI >=40:
  print(Name +" " +  "Morbidly obese")
  print("The Health Risk is Very High")

else:
  print(Name +" " +  "Entered invalid input")
```
